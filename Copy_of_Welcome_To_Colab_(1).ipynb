#Univariate analysis




Batting Team Distribution

```
plt.figure(figsize=(12, 6))

df["batting_team"].value_counts().plot(kind="bar")

plt.title("Distribution of Batting Teams")
plt.xlabel("Batting Team")
plt.ylabel("Number of Deliveries")

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

##Year wise batsman distribution






```
import matplotlib.pyplot as plt

yearly_runs = (df.groupby(["year", "batter"])["runs_batter"].sum().reset_index())
top_batsmen = yearly_runs.loc[yearly_runs.groupby("year")["runs_batter"].idxmax()].sort_values("year")
plt.figure(figsize=(18, 8))
bars = plt.bar(top_batsmen["year"].astype(str),top_batsmen["runs_batter"])

for bar, batsman in zip(bars, top_batsmen["batter"]):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 10,
        batsman,
        ha="center",
        rotation=90,
        fontsize=8
    )

plt.title("Year-wise Highest Run Scorer in IPL")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

##Year-wise Highest Run Scoring Team

```
yearly_team_runs = (
    df.groupby(["year", "batting_team"])["runs_total"]
      .sum()
)

top = yearly_team_runs.groupby(level=0).idxmax()

years = [i[0] for i in top]
teams = [i[1] for i in top]
runs = [yearly_team_runs[i] for i in top]

plt.figure(figsize=(18, 8))  # chart moto karo

bars = plt.bar(years, runs)

# Team name ne horizontal rakho
for bar, team in zip(bars, teams):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 100,
        team,
        ha="center",
        va="bottom",
        fontsize=7,
        rotation=45
    )

plt.title("Year-wise Highest Run Scoring Team")
plt.xlabel("Year")
plt.ylabel("Total Runs")
plt.xticks(years)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#Bar plot using Venue vs Team Runs vs Batting Team

#Scatter plot using Batter Balls vs Batter Runs by Batting Position

```
sample_data = df[
    ["batter_balls", "batter_runs", "bat_pos"]
].dropna()

sample_data = sample_data.sample(
    min(3000, len(sample_data)),
    random_state=42
)

plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=sample_data,
    x="batter_balls",
    y="batter_runs",
    hue="bat_pos",
    s=70
)

plt.title("Batter Runs vs Balls Faced by Batting Position")
plt.xlabel("Balls Faced")
plt.ylabel("Batter Runs")

plt.tight_layout()
plt.show()
```

#Bar plot using Batting Team vs Over Phase vs Runs

```
df["over_phase"] = pd.cut(
    df["over"],
    bins=[-1, 5, 14, 19],
    labels=[
        "Powerplay",
        "Middle Overs",
        "Death Overs"
    ]
)

phase_data = df.groupby(
    ["batting_team", "over_phase"],
    observed=True
)["runs_total"].mean().reset_index()

plt.figure(figsize=(15, 7))

sns.barplot(
    data=phase_data,
    x="batting_team",
    y="runs_total",
    hue="over_phase"
)

plt.title("Team Scoring Across Match Phases")
plt.xlabel("Batting Team")
plt.ylabel("Average Runs per Delivery")

plt.xticks(rotation=60)

plt.tight_layout()
plt.show()
```

#Heatmap using Bowling Team vs Batting Team vs Team Runs

```
match_team_data = df.groupby(
    ["batting_team", "bowling_team"]
)["team_runs"].mean().reset_index()

matchup_pivot = match_team_data.pivot(
    index="bowling_team",
    columns="batting_team",
    values="team_runs"
)

plt.figure(figsize=(13, 10))

sns.heatmap(
    matchup_pivot,
    annot=True,
    fmt=".1f"
)

plt.title("Average Team Runs by Batting Team vs Bowling Team")
plt.xlabel("Batting Team")
plt.ylabel("Bowling Team")

plt.tight_layout()
plt.show()
```

#Scatter plot using Team Runs vs Team Wickets by Year

```
year_performance = df.groupby(
    "year"
).agg(
    average_runs=("team_runs", "mean"),
    average_wickets=("team_wicket", "mean")
).reset_index()

plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=year_performance,
    x="average_wickets",
    y="average_runs",
    hue="year",
    s=150
)

plt.title("Average Team Runs vs Average Wickets Across Years")
plt.xlabel("Average Wickets")
plt.ylabel("Average Team Runs")

plt.tight_layout()
plt.show()
```

#Heatmap using Correlation Matrix of Numeric Features

```
numeric_columns = [
    "team_runs",
    "team_balls",
    "team_wicket",
    "batter_runs",
    "batter_balls",
    "runs_total",
    "runs_bowler"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap of IPL Numeric Features")
plt.xlabel("Numeric Features")
plt.ylabel("Numeric Features")

plt.tight_layout()
plt.show()
```
